# Performance Analytics

Computes daily returns, CAGR, Sharpe, Sortino, alpha, beta, maximum drawdown, composite fund score, and benchmark comparison.

```
from pathlib import Path
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import linregress

PROJECT_ROOT = Path('..').resolve()
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'
nav = pd.read_csv(PROCESSED_DIR / 'nav_history_cleaned.csv', parse_dates=['date'])
fund = pd.read_csv(PROCESSED_DIR / 'fund_master_cleaned.csv')
benchmark = pd.read_csv(PROCESSED_DIR / 'benchmark_indices_cleaned.csv', parse_dates=['date'])
scorecard = pd.read_csv(PROCESSED_DIR / 'fund_scorecard.csv')
alpha_beta = pd.read_csv(PROCESSED_DIR / 'alpha_beta.csv')
```

Daily returns are computed as `nav_t / nav_t-1 - 1` for each AMFI scheme and stored in `daily_returns.csv`.

```
daily_returns = pd.read_csv(PROCESSED_DIR / 'daily_returns.csv', parse_dates=['date'])
daily_returns['daily_return'].describe()
```

CAGR is calculated over 1-year, 3-year, and 5-year windows using the latest NAV date as the endpoint.

```
scorecard[['scheme_name', 'cagr_1yr', 'cagr_3yr', 'cagr_5yr']].head(10)
```

Sharpe and Sortino use a 6.5% annual risk-free rate proxy.

```
scorecard[['scheme_name', 'sharpe_ratio_calc', 'sortino_ratio_calc']].head(10)
```

Alpha and beta are estimated by OLS regression of fund daily returns against Nifty 100 returns.

```
alpha_beta.sort_values('alpha', ascending=False).head(10)
```

Maximum drawdown records the worst peak-to-trough decline and the corresponding start/trough dates.

```
scorecard[['scheme_name', 'max_drawdown', 'drawdown_start', 'drawdown_trough']].sort_values('max_drawdown').head(10)
```

The 0-100 fund score combines 3-year CAGR, Sharpe, alpha, inverse expense ratio, and inverse drawdown rank.

```
scorecard[['scheme_name', 'fund_score', 'cagr_3yr', 'sharpe_ratio_calc', 'alpha', 'expense_ratio_pct', 'max_drawdown']].head(10)
```
